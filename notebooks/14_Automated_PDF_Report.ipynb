```python
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    PageBreak
)

from reportlab.lib.styles import (
    getSampleStyleSheet
)

import pandas as pd

print(
    "Libraries Loaded Successfully"
)
```

```python
executive_dashboard = pd.read_csv(
    'executive_dashboard.csv'
)

executive_kpis = pd.read_csv(
    'executive_kpis.csv'
)

customer_360 = pd.read_csv(
    'customer_360.csv'
)

marketing_customers = pd.read_csv(
    'marketing_intelligence_customers.csv'
)

print(
    "Datasets Loaded Successfully"
)
```

# Executive Summary Generator

```python
business_score = (
    executive_dashboard[
        executive_dashboard[
            'Metric'
        ]
        ==
        'Business Score'
    ]['Value']
    .iloc[0]
)

executive_summary = f"""

Customer Intelligence Platform Report

Business Intelligence Score:
{round(business_score,2)}

The business demonstrates a data-driven customer intelligence framework with customer segmentation, churn prediction, marketing intelligence and executive decision support capabilities.

"""
```

# Business Performance Summary

```python
business_summary = f"""

Business Performance Summary

Average Customer Health:
{round(customer_360['Customer_Health_Index'].mean(),2)}

Average Strategic Value:
{round(customer_360['Strategic_Value_Score'].mean(),2)}

Average Risk Score:
{round(customer_360['Dynamic_Risk_Score'].mean(),2)}

"""
```

# Customer Intelligence Summary

```python
customer_summary = f"""

Customer Intelligence Summary

Total Customers:
{len(customer_360)}

Top Personas:

{customer_360['Customer_Persona'].value_counts().head()}

"""
```

# Marketing Intelligence Summary

```python
marketing_summary = f"""

Marketing Intelligence Summary

Total Marketing Profiles:
{len(marketing_customers)}

Marketing Priority Distribution:

{marketing_customers['Marketing_Priority'].value_counts()}

"""
```

# Executive Recommendations

```python
recommendations = """

Executive Recommendations

1. Improve retention of high-risk customers.

2. Focus on high-value customer segments.

3. Expand loyalty programmes.

4. Prioritise strategic customers.

5. Continue monitoring customer health metrics.

"""
```

# Build PDF

```python
doc = SimpleDocTemplate(
    "Customer_Intelligence_Report.pdf"
)

styles = getSampleStyleSheet()

elements = []
```

```python
elements.append(
    Paragraph(
        "Customer Intelligence Platform",
        styles['Title']
    )
)

elements.append(
    Spacer(1,12)
)

elements.append(
    Paragraph(
        executive_summary,
        styles['BodyText']
    )
)

elements.append(
    Spacer(1,12)
)

elements.append(
    Paragraph(
        business_summary,
        styles['BodyText']
    )
)

elements.append(
    Spacer(1,12)
)

elements.append(
    Paragraph(
        customer_summary,
        styles['BodyText']
    )
)

elements.append(
    Spacer(1,12)
)

elements.append(
    Paragraph(
        marketing_summary,
        styles['BodyText']
    )
)

elements.append(
    Spacer(1,12)
)

elements.append(
    Paragraph(
        recommendations,
        styles['BodyText']
    )
)
```

```python
doc.build(
    elements
)

print(
    "PDF Report Generated Successfully"
)
```
