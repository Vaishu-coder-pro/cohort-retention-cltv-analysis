# Import Libraries

```python
import pandas as pd
import numpy as np

print("Libraries Loaded Successfully")
```

# Load Cleaned Dataset

```python
master = pd.read_csv(
    'cleaned_master_dataset.csv'
)

print("Dataset Loaded Successfully")
```

# Check Dataset

```python
master.shape
```

# Top 10 Revenue Categories

```python
category_revenue = (
    master.groupby(
        'product_category_name_english'
    )['revenue']
    .sum()
    .sort_values(
        ascending=False
    )
)

category_revenue.head(10)
```

# Top 10 States by Revenue

```python
state_revenue = (
    master.groupby(
        'customer_state'
    )['revenue']
    .sum()
    .sort_values(
        ascending=False
    )
)

state_revenue.head(10)
```

# Most Used Payment Types

```python
master[
    'payment_type'
].value_counts()
```

# Average Delivery Days

```python
master[
    'delivery_days'
].mean()
```

# Delivery Status Distribution

```python
master[
    'delivery_status'
].value_counts()
```

# Save KPI Summary

```python
kpi_summary = pd.DataFrame({

    'Metric':[
        'Total Revenue',
        'Total Customers',
        'Average Delivery Days'
    ],

    'Value':[

        master['revenue'].sum(),

        master['customer_unique_id'].nunique(),

        master['delivery_days'].mean()
    ]
})

kpi_summary
```
