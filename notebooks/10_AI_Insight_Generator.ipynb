# Customer Risk Intelligence

```python
risk_summary = (
    customer_churn[
        'Risk_Level'
    ]
    .value_counts()
)

print(
    "Customer Risk Distribution:"
)

print(risk_summary)
```

```python
critical_count = risk_summary.get(
    'Critical Risk',
    0
)

if critical_count > 10000:

    print(
        "AI Recommendation: Immediate retention campaigns should be launched for critical-risk customers."
    )

else:

    print(
        "AI Recommendation: Customer retention is currently stable."
    )
```

# Cluster Intelligence Engine

# Revenue Forecast Intelligence

```python
forecast_avg = (
    forecast_summary[
        'Value'
    ]
    .mean()
)

print(
    f"Average forecast indicator value: {round(forecast_avg,2)}"
)
```

```python
churn_rate = (
    customer_churn['Churn']
    .mean()
    * 100
)

print(
    "Churn Rate:",
    round(churn_rate, 2),
    "%"
)
```

```python
historical_revenue = (
    forecast_summary[
        forecast_summary['Metric']
        ==
        'Average Historical Revenue'
    ]['Value']
    .iloc[0]
)

forecast_revenue = (
    forecast_summary[
        forecast_summary['Metric']
        ==
        'Average Forecast Revenue'
    ]['Value']
    .iloc[0]
)

forecast_growth = (
    (
        forecast_revenue
        -
        historical_revenue
    )
    /
    historical_revenue
) * 100

print(
    "Forecast Growth:",
    round(forecast_growth, 2),
    "%"
)
```

# Business Health Score

```python
health_score = 100

if critical_count > 10000:
    health_score -= 25

if churn_rate > 70:
    health_score -= 25

if forecast_growth < 0:
    health_score -= 25

print(
    "Business Health Score:",
    health_score,
    "/100"
)
```

```python
health_score = 100

if critical_count > 10000:
    health_score -= 20

if churn_rate > 70:
    health_score -= 20

if forecast_growth < 0:
    health_score -= 20

if top_category['revenue'] < 100000:
    health_score -= 20

health_score = max(
    health_score,
    0
)

print(
    "Business Health Score:",
    health_score,
    "/100"
)
```
