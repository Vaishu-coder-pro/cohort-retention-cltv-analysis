```python
customer_360 = pd.read_csv(
    'customer_360.csv'
)

executive_dashboard = pd.read_csv(
    'executive_dashboard.csv'
)

print(
    "Datasets Loaded Successfully"
)
```

# Customer Health Anomaly Detection

```python
health_threshold = (

    customer_360[
        'Customer_Health_Index'
    ].mean()

    -

    2 *

    customer_360[
        'Customer_Health_Index'
    ].std()

)

health_anomalies = (

    customer_360[
        customer_360[
            'Customer_Health_Index'
        ] < health_threshold
    ]

)

print(
    "Health Anomalies:",
    len(
        health_anomalies
    )
)
```

# Risk Anomaly Detection

```python
risk_threshold = (

    customer_360[
        'Dynamic_Risk_Score'
    ].mean()

    +

    2 *

    customer_360[
        'Dynamic_Risk_Score'
    ].std()

)

risk_anomalies = (

    customer_360[
        customer_360[
            'Dynamic_Risk_Score'
        ] > risk_threshold
    ]

)

print(
    "Risk Anomalies:",
    len(
        risk_anomalies
    )
)
```

# Strategic Value Anomaly Detection

```python
value_threshold = (

    customer_360[
        'Strategic_Value_Score'
    ].mean()

    +

    2 *

    customer_360[
        'Strategic_Value_Score'
    ].std()

)

value_anomalies = (

    customer_360[
        customer_360[
            'Strategic_Value_Score'
        ] > value_threshold
    ]

)

print(
    "Strategic Value Anomalies:",
    len(
        value_anomalies
    )
)
```

# Anomaly Summary

```python
anomaly_summary = pd.DataFrame({

    'Anomaly_Type':[

        'Health',

        'Risk',

        'Strategic Value'

    ],

    'Count':[

        len(
            health_anomalies
        ),

        len(
            risk_anomalies
        ),

        len(
            value_anomalies
        )

    ]

})

anomaly_summary
```

# Business Alert Engine

```python
alerts = []

if len(health_anomalies) > 1000:

    alerts.append(
        "Customer health decline detected."
    )

if len(risk_anomalies) > 1000:

    alerts.append(
        "Customer risk surge detected."
    )

if len(value_anomalies) > 500:

    alerts.append(
        "High-value customer concentration detected."
    )

for alert in alerts:

    print(
        "ALERT:",
        alert
    )
```

# Anomaly Risk Level

```python
total_anomalies = (

    len(
        health_anomalies
    )

    +

    len(
        risk_anomalies
    )

    +

    len(
        value_anomalies
    )

)

if total_anomalies > 5000:

    anomaly_risk = (
        "High"
    )

elif total_anomalies > 1000:

    anomaly_risk = (
        "Medium"
    )

else:

    anomaly_risk = (
        "Low"
    )

print(
    "Business Anomaly Risk:",
    anomaly_risk
)
```

```python
health_anomalies.to_csv(
    'health_anomalies.csv',
    index=False
)

risk_anomalies.to_csv(
    'risk_anomalies.csv',
    index=False
)

value_anomalies.to_csv(
    'value_anomalies.csv',
    index=False
)

anomaly_summary.to_csv(
    'anomaly_summary.csv',
    index=False
)

print(
    "Anomaly Datasets Exported Successfully"
)
```

# Anomaly Severity Score

```python
anomaly_summary['Severity_Score'] = (

    anomaly_summary['Count']

    /

    anomaly_summary['Count'].max()

) * 100

anomaly_summary
```

# Anomaly Severity Classification

```python
anomaly_summary[
    'Severity_Level'
] = np.where(

    anomaly_summary[
        'Severity_Score'
    ] > 75,

    'Critical',

    np.where(

        anomaly_summary[
            'Severity_Score'
        ] > 40,

        'Moderate',

        'Low'

    )

)

anomaly_summary
```

# Executive Alert Dashboard

```python
executive_alerts = []

for _, row in anomaly_summary.iterrows():

    if row[
        'Severity_Level'
    ] == 'Critical':

        executive_alerts.append(

            f"{row['Anomaly_Type']} anomaly requires immediate investigation."

        )

for alert in executive_alerts:

    print(
        alert
    )
```

# Predictive Anomaly Risk Score

```python
predictive_risk = (

    anomaly_summary[
        'Severity_Score'
    ].mean()

)

print(
    "Predictive Anomaly Risk:",
    round(
        predictive_risk,
        2
    )
)
```
