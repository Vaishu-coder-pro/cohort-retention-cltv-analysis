```python
import pandas as pd
import numpy as np

print("Libraries Loaded Successfully")
```

```python
customer_360 = pd.read_csv(
    'customer_360.csv'
)

product_category = pd.read_csv(
    'product_category_revenue.csv'
)

customer_churn = pd.read_csv(
    'customer_churn.csv'
)

print(
    "Datasets Loaded Successfully"
)
```

# Campaign Recommendation Engine

```python
def campaign_type(persona):

    if persona == 'Elite VIP':
        return 'Premium Rewards Campaign'

    elif persona == 'VIP Customer':
        return 'Loyalty Programme Campaign'

    elif persona == 'Loyal Customer':
        return 'Customer Appreciation Campaign'

    elif persona == 'At Risk Customer':
        return 'Retention Campaign'

    elif persona == 'Inactive Customer':
        return 'Win Back Campaign'

    else:
        return 'Standard Promotion'


customer_360[
    'Recommended_Campaign'
] = customer_360[
    'Customer_Persona'
].apply(
    campaign_type
)

customer_360[
    [
        'Customer_Persona',
        'Recommended_Campaign'
    ]
].head()
```

# Customer Targeting Engine

```python
target_customers = (
    customer_360
    .sort_values(
        'Strategic_Value_Score',
        ascending=False
    )
)

target_customers[
    [
        'customer_unique_id',
        'Customer_Persona',
        'Strategic_Value_Score'
    ]
].head(20)
```

# Revenue Leakage Detector

```python
lowest_categories = (
    product_category
    .sort_values(
        'revenue'
    )
    .head(5)
)

lowest_categories
```

```python
print(
    "AI Insight:"
)

print(
    "These categories require pricing, marketing or inventory review."
)
```

# Revenue Opportunity Finder

```python
top_categories = (
    product_category
    .sort_values(
        'revenue',
        ascending=False
    )
    .head(5)
)

top_categories
```

```python
print(
    "AI Insight:"
)

print(
    "These categories represent the strongest growth opportunities."
)
```

# Upsell Intelligence Engine

```python
upsell_customers = (
    customer_360[
        customer_360[
            'Customer_Persona'
        ].isin(
            [
                'VIP Customer',
                'Elite VIP'
            ]
        )
    ]
)

upsell_customers[
    [
        'customer_unique_id',
        'Customer_Persona'
    ]
].head(20)
```

# Retention Campaign Generator

```python
retention_customers = (
    customer_360[
        customer_360[
            'Retention_Priority_Score'
        ] > 70
    ]
)

retention_customers[
    [
        'customer_unique_id',
        'Retention_Priority_Score'
    ]
].head(20)
```

```python
retention_customers[
    'Retention_Campaign'
] = (
    'Special Discount Offer'
)
```

```python
retention_customers = (
    retention_customers.copy()
)

retention_customers.loc[
    :,
    'Retention_Campaign'
] = 'Special Discount Offer'
```

# Marketing ROI Priority Matrix

```python
customer_360[
    'Marketing_Priority'
] = np.where(

    customer_360[
        'Strategic_Value_Score'
    ] > 80,

    'High',

    'Normal'

)

customer_360[
    'Marketing_Priority'
].value_counts()
```

# Marketing Health Index

```python
marketing_health = 100

if len(retention_customers) > 10000:
    marketing_health -= 20

if len(lowest_categories) > 3:
    marketing_health -= 20

print(
    "Marketing Health Index:",
    marketing_health,
    "/100"
)
```

# Customer Conversion Score

```python
customer_360[
    'Conversion_Score'
] = (

    customer_360[
        'Customer_Health_Index'
    ] * 0.5

    +

    customer_360[
        'Strategic_Value_Score'
    ] * 0.5

)

customer_360[
    [
        'customer_unique_id',
        'Conversion_Score'
    ]
].head()
```

# Campaign Priority Ranking

```python
campaign_priority = (
    customer_360
    .sort_values(
        'Conversion_Score',
        ascending=False
    )
)

campaign_priority.head(20)
```

# Customer Acquisition Intelligence

```python
potential_targets = (
    customer_360[
        customer_360[
            'Conversion_Score'
        ] > 75
    ]
)

print(
    "High Potential Marketing Targets:"
)

print(
    len(potential_targets)
)
```

# Marketing Opportunity Score

```python
opportunity_score = (

    top_categories[
        'revenue'
    ].sum()

    /

    product_category[
        'revenue'
    ].sum()

) * 100

print(
    "Marketing Opportunity Score:",
    round(opportunity_score,2),
    "%"
)
```
