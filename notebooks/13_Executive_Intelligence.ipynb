```python
import pandas as pd
import numpy as np

print(
    "Libraries Loaded Successfully"
)
```

```python
customer_360 = pd.read_csv(
    'customer_360.csv'
)

forecast_summary = pd.read_csv(
    'forecast_summary.csv'
)

marketing_customers = pd.read_csv(
    'marketing_intelligence_customers.csv'
)

print(
    "Executive Datasets Loaded"
)
```

# Business Intelligence Score

```python
health_score = (
    customer_360[
        'Customer_Health_Index'
    ].mean()
)

risk_score = (
    100
    -
    customer_360[
        'Dynamic_Risk_Score'
    ].mean()
)

business_score = (

    health_score * 0.5

    +

    risk_score * 0.5

)

print(
    "Business Intelligence Score:",
    round(
        business_score,
        2
    ),
    "/100"
)
```

# Executive KPI Scorecard

```python
executive_kpis = pd.DataFrame({

    'Metric': [

        'Business Score',

        'Average Customer Health',

        'Average Strategic Value',

        'Average Risk Score'

    ],

    'Value': [

        round(
            business_score,
            2
        ),

        round(
            customer_360[
                'Customer_Health_Index'
            ].mean(),
            2
        ),

        round(
            customer_360[
                'Strategic_Value_Score'
            ].mean(),
            2
        ),

        round(
            customer_360[
                'Dynamic_Risk_Score'
            ].mean(),
            2
        )

    ]

})

executive_kpis
```

# Strategic Risk Analysis

```python
high_risk = (
    customer_360[
        customer_360[
            'Dynamic_Risk_Score'
        ] > 75
    ]
)

print(
    "High Risk Customers:",
    len(high_risk)
)
```

# Strategic Opportunity Analysis

```python
top_value = (
    customer_360
    .sort_values(
        'Strategic_Value_Score',
        ascending=False
    )
)

top_value.head(10)
```

# Business Maturity Assessment

```python
if business_score >= 80:

    maturity = "Advanced"

elif business_score >= 60:

    maturity = "Growing"

else:

    maturity = "Developing"

print(
    "Business Maturity:",
    maturity
)
```

# Executive Decision Engine

```python
decisions = []

if len(high_risk) > 10000:

    decisions.append(
        "Launch customer retention programme."
    )

decisions.append(
    "Expand high-value customer initiatives."
)

decisions.append(
    "Increase loyalty programme investment."
)

for decision in decisions:

    print(
        decision
    )
```

# Board Meeting Summary

```python
board_summary = f"""

BOARD MEETING SUMMARY

Business Score:
{round(business_score,2)}

Business Maturity:
{maturity}

High Risk Customers:
{len(high_risk)}

Recommended Actions:

1. Improve customer retention.

2. Focus on strategic customers.

3. Increase customer engagement.

"""

print(
    board_summary
)
```

# CEO Action Plan

```python
ceo_actions = []

if len(high_risk) > 10000:
    ceo_actions.append(
        "Immediate customer retention initiative required."
    )

if business_score < 70:
    ceo_actions.append(
        "Improve overall customer health and engagement."
    )

ceo_actions.append(
    "Increase investment in high-value customers."
)

ceo_actions.append(
    "Expand loyalty and retention programmes."
)

ceo_actions.append(
    "Monitor strategic business KPIs monthly."
)

for action in ceo_actions:

    print(
        "CEO Action:",
        action
    )
```

# Executive Priority Matrix

```python
customer_360[
    'Executive_Priority'
] = np.where(

    customer_360[
        'Strategic_Value_Score'
    ] > 75,

    'High Priority',

    'Normal Priority'

)

customer_360[
    'Executive_Priority'
].value_counts()
```

# Business Stability Index

```python
stability_index = (

    customer_360[
        'Customer_Health_Index'
    ].mean()

    -

    customer_360[
        'Dynamic_Risk_Score'
    ].mean()

    + 100

)

print(
    "Business Stability Index:",
    round(
        stability_index,
        2
    )
)
```

```python
if stability_index > 80:

    print(
        "Business Stability: Strong"
    )

elif stability_index > 60:

    print(
        "Business Stability: Moderate"
    )

else:

    print(
        "Business Stability: Weak"
    )
```

# Strategic Growth Index

```python
growth_index = (

    customer_360[
        'Strategic_Value_Score'
    ].mean()

    *

    business_score

) / 100

print(
    "Strategic Growth Index:",
    round(
        growth_index,
        2
    )
)
```

# Executive Dashboard Dataset

```python
executive_dashboard = pd.DataFrame({

    'Metric':[

        'Business Score',

        'Business Stability Index',

        'Strategic Growth Index',

        'High Risk Customers'

    ],

    'Value':[

        round(
            business_score,
            2
        ),

        round(
            stability_index,
            2
        ),

        round(
            growth_index,
            2
        ),

        len(
            high_risk
        )

    ]

})

executive_dashboard
```

```python
if business_score >= 85:

    executive_status = (
        "Excellent"
    )

elif business_score >= 70:

    executive_status = (
        "Healthy"
    )

elif business_score >= 50:

    executive_status = (
        "Needs Improvement"
    )

else:

    executive_status = (
        "Critical"
    )

print(
    "Executive Status:",
    executive_status
)
```

```python
executive_dashboard.to_csv(
    'executive_dashboard.csv',
    index=False
)

executive_kpis.to_csv(
    'executive_kpis.csv',
    index=False
)

print(
    "Executive Intelligence Datasets Exported Successfully"
)
```
