```python
import pandas as pd
import numpy as np

print("Libraries Loaded Successfully")
```

```python
master = pd.read_csv(
    'cleaned_master_dataset.csv'
)

print("Dataset Loaded Successfully")
```

```python
master.shape
```

# Check available product columns

```python
[
    c for c in master.columns
    if 'product' in c.lower()
]
```

# Top Revenue Categories

```python
category_revenue = (
    master
    .groupby(
        'product_category_name_english'
    )['revenue']
    .sum()
    .reset_index()
    .sort_values(
        'revenue',
        ascending=False
    )
)

category_revenue.head(10)
```

# Most Sold Categories

```python
top_categories = (
    master
    .groupby(
        'product_category_name_english'
    )
    .size()
    .reset_index(
        name='Total_Sales'
    )
    .sort_values(
        'Total_Sales',
        ascending=False
    )
)

top_categories.head(10)
```

# Top Revenue Products

```python
master[
[
'product_id',
'revenue'
]
].sort_values(
    'revenue',
    ascending=False
).head(10)
```

# Product KPIs

```python
print(
    "Average Revenue:",
    round(master['revenue'].mean(),2)
)

print(
    "Maximum Revenue:",
    round(master['revenue'].max(),2)
)

print(
    "Minimum Revenue:",
    round(master['revenue'].min(),2)
)
```

```python
monthly_revenue = (
    master
    .groupby(
        [
            'purchase_year',
            'purchase_month'
        ]
    )['revenue']
    .sum()
    .reset_index()
)

monthly_revenue.head()
```

```python
category_revenue.to_csv(
    'product_category_revenue.csv',
    index=False
)

monthly_revenue.to_csv(
    'monthly_revenue.csv',
    index=False
)

top_categories.to_csv(
    'top_categories.csv',
    index=False
)

print("Files Saved Successfully")
```
