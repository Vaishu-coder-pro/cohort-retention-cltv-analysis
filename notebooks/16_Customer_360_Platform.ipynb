```python
import pandas as pd
import numpy as np

print(
    "Libraries Loaded Successfully"
)
```

```python
customer_360 = pd.read_csv(
    'customer_360.csv'
)

marketing_customers = pd.read_csv(
    'marketing_intelligence_customers.csv'
)

customer_predictions = pd.read_csv(
    'customer_churn_predictions.csv'
)

print(
    "Datasets Loaded Successfully"
)
```

# Create Customer 360 Master Dataset

```python
customer_360_master = pd.merge(

    customer_360,

    customer_predictions[
        [
            'customer_unique_id',
            'Churn_Probability'
        ]
    ],

    on='customer_unique_id',

    how='left'

)

customer_360_master.head()
```

# Customer Ranking Engine

```python
customer_360_master[
    'Customer_Rank'
] = (

    customer_360_master[
        'Strategic_Value_Score'
    ]

    .rank(
        ascending=False
    )

)

customer_360_master[
    [
        'customer_unique_id',
        'Customer_Rank'
    ]
].head()
```

# Customer Tier Classification

```python
conditions = [

    customer_360_master[
        'Strategic_Value_Score'
    ] >= 80,

    customer_360_master[
        'Strategic_Value_Score'
    ] >= 60,

    customer_360_master[
        'Strategic_Value_Score'
    ] >= 40

]

choices = [

    'Platinum',

    'Gold',

    'Silver'

]

customer_360_master[
    'Customer_Tier'
] = np.select(

    conditions,

    choices,

    default='Bronze'

)

customer_360_master[
    'Customer_Tier'
].value_counts()
```

# Strategic Customer Identification

```python
strategic_customers = (

    customer_360_master[

        customer_360_master[
            'Strategic_Value_Score'
        ] > 80

    ]

)

print(
    "Strategic Customers:",
    len(
        strategic_customers
    )
)
```

# Customer Value Matrix

```python
value_matrix = pd.DataFrame({

    'Metric':[

        'Average CLTV',

        'Average Health Index',

        'Average Strategic Value',

        'Average Risk Score'

    ],

    'Value':[

        round(
            customer_360_master[
                'CLTV'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Customer_Health_Index'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Strategic_Value_Score'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Dynamic_Risk_Score'
            ].mean(),
            2
        )

    ]

})

value_matrix
```

# Customer Summary Dashboard

```python
summary_dashboard = pd.DataFrame({

    'Metric':[

        'Total Customers',

        'Strategic Customers',

        'Average Health Index',

        'Average CLTV'

    ],

    'Value':[

        len(
            customer_360_master
        ),

        len(
            strategic_customers
        ),

        round(
            customer_360_master[
                'Customer_Health_Index'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'CLTV'
            ].mean(),
            2
        )

    ]

})

summary_dashboard
```

# AI Customer Advisor

```python
print(
    "AI CUSTOMER ADVISOR"
)

print(
    "---------------------"
)

print(
    f"Total Customers: {len(customer_360_master)}"
)

print(
    f"Strategic Customers: {len(strategic_customers)}"
)

print(
    "Focus on Platinum and Gold customers."
)

print(
    "Protect high-value customer segments."
)

print(
    "Prioritise customers with high retention priority."
)
```

# Customer Portfolio Score

```python
portfolio_score = (

    customer_360_master[
        'Strategic_Value_Score'
    ].mean()

    *

    customer_360_master[
        'Customer_Health_Index'
    ].mean()

) / 100

print(
    "Customer Portfolio Score:",
    round(
        portfolio_score,
        2
    )
)
```

```python
customer_360_master.to_csv(
    'customer_360_master.csv',
    index=False
)

summary_dashboard.to_csv(
    'customer_360_summary.csv',
    index=False
)

value_matrix.to_csv(
    'customer_value_matrix.csv',
    index=False
)

print(
    "Customer 360 Platform Exported Successfully"
)
```

# Customer Lifetime Tier Score

```python
customer_360_master['Lifetime_Tier_Score'] = (

    customer_360_master['CLTV'] * 0.4

    +

    customer_360_master['Customer_Health_Index'] * 0.3

    +

    customer_360_master['Strategic_Value_Score'] * 0.3

)

customer_360_master[
    'Lifetime_Tier_Score'
] = customer_360_master[
    'Lifetime_Tier_Score'
].round(2)
```

# Customer Engagement Score

```python
customer_360_master[
    'Engagement_Score'
] = (

    customer_360_master[
        'Customer_Health_Index'
    ] * 0.6

    +

    (
        100
        -
        customer_360_master[
            'Dynamic_Risk_Score'
        ]
    ) * 0.4

)

customer_360_master[
    'Engagement_Score'
] = customer_360_master[
    'Engagement_Score'
].round(2)
```

# Customer Profitability Index

```python
customer_360_master[
    'Profitability_Index'
] = (

    customer_360_master[
        'CLTV'
    ]

    /

    customer_360_master[
        'CLTV'
    ].max()

) * 100

customer_360_master[
    'Profitability_Index'
] = customer_360_master[
    'Profitability_Index'
].round(2)
```

# Customer Loyalty Score

```python
customer_360_master[
    'Loyalty_Score'
] = (

    customer_360_master[
        'Customer_Health_Index'
    ] * 0.7

    +

    customer_360_master[
        'Strategic_Value_Score'
    ] * 0.3

)

customer_360_master[
    'Loyalty_Score'
] = customer_360_master[
    'Loyalty_Score'
].round(2)
```

# Customer Investment Priority

```python
conditions = [

    customer_360_master[
        'Strategic_Value_Score'
    ] > 80,

    customer_360_master[
        'Strategic_Value_Score'
    ] > 60

]

choices = [

    'High Investment',

    'Medium Investment'

]

customer_360_master[
    'Investment_Priority'
] = np.select(

    conditions,

    choices,

    default='Low Investment'

)
```

# Top 100 Strategic Customers

```python
top100 = (

    customer_360_master
    .sort_values(
        'Strategic_Value_Score',
        ascending=False
    )
    .head(100)

)

top100.to_csv(
    'top_100_strategic_customers.csv',
    index=False
)
```

```python
portfolio_health = pd.DataFrame({

    'Metric':[

        'Average Health',

        'Average Loyalty',

        'Average Profitability',

        'Average Engagement'

    ],

    'Value':[

        round(
            customer_360_master[
                'Customer_Health_Index'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Loyalty_Score'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Profitability_Index'
            ].mean(),
            2
        ),

        round(
            customer_360_master[
                'Engagement_Score'
            ].mean(),
            2
        )

    ]

})

portfolio_health
```
